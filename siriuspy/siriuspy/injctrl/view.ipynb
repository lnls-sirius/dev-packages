```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
def delta_current(topup_period=1, nr_pulses=1, target_current=200, lifetime=9 * 3600): # NOQA
    frac = (topup_period * 60) / lifetime
    dcurr = target_current * frac
    dcurr /= nr_pulses
    d1 = dcurr / (1 + frac)
    return dcurr, d1
```

```python
lt = 7 * 3600
```

```python
t = np.linspace(0, 7, 60) * 3600
plt.plot(t, 200*np.exp(-t/lt))
```
