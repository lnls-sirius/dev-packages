```python
import time as _time

import numpy as _np
import matplotlib.pyplot as plt

from siriuspy.oscilloscope import Keysight, Scopes, ScopeSignals, WaveformAcquisition

from siriuspy.currinfo import LICurrInfoApp
```

```python
ScopeSignals.get_signals(Scopes.LI_DI_ICTOSC)
```

```python
def plot_waveforms(wfms):
    for chan in wfms:
        plt.plot(1e6*wfms[chan][0], 1e3*wfms[chan][1], label=chan)
    plt.xlabel('time [µs]')
    plt.ylabel('Voltage [mV]')
    plt.grid()
    plt.legend()
    plt.tight_layout()
    plt.show()

def process_waveform_0(wfm):
    tim, val = wfm

    indcs = _np.arange(len(val))
    max_idx = _np.argmax(val)
    tim_peak = tim[max_idx]
    sel1 = (val - val[0]) < perc * (val[max_idx] - val[0])
    sel2 = indcs < max_idx
    sel = sel1 & sel2

    pcoeffs = (0, 0)
    val_fit = 0 * val
    
    tim_fix = tim
    val_fix = val - val_fit

    charge = 1e9 * _np.trapz(val_fix, tim_fix) / 5
    params = charge, tim, val, sel, val_fit, tim_fix, val_fix, tim_peak, pcoeffs
    return params
    
def process_waveform_1(wfm, perc, order):
    tim, val = wfm

    indcs = _np.arange(len(val))
    max_idx = _np.argmax(val)
    tim_peak = tim[max_idx]
    sel1 = (val - val[0]) < perc * (val[max_idx] - val[0])
    sel2 = indcs < max_idx
    sel = sel1 & sel2

    pcoeffs = _np.polynomial.polynomial.polyfit(tim[sel], val[sel], order)
    val_fit = _np.polynomial.polynomial.polyval(tim, pcoeffs)
    
    tim_fix = tim
    val_fix = val - val_fit

    charge = 1e9 * _np.trapz(val_fix, tim_fix) / 5
    params = charge, tim, val, sel, val_fit, tim_fix, val_fix, tim_peak, pcoeffs
    return params

def process_waveform_2(wfm, perc, order):
    tim, val = wfm

    indcs = _np.arange(len(val))
    max_idx = _np.argmax(val)
    tim_peak = tim[max_idx]
    sel1 = (val - val[0]) < perc * (val[max_idx] - val[0])
    sel2 = indcs < max_idx
    sel = sel1 & sel2

    pcoeffs = _np.polynomial.polynomial.polyfit(tim[sel], val[sel], order)
    val_fit = _np.polynomial.polynomial.polyval(tim, pcoeffs)

    sel1 = (val < val_fit) & (tim > tim[max_idx])
    sel2 = tim < min(tim[sel1])
    
    tim_fix = tim[sel2]
    val_fix = (val - val_fit)[sel2]

    charge = 1e9 * _np.trapz(val_fix, tim_fix) / 5
    params = charge, tim, val, sel, val_fit, tim_fix, val_fix, tim_peak, pcoeffs
    return params

    
def analyse_waveform(func, wfm, process_params, plot_flag):
    
    params = func(wfm, *process_params)
    charge, tim, val, sel, val_fit, tim_fix, val_fix, tim_peak, pcoeffs = params

    if plot_flag:
        plt.plot(1e6*tim, 1e3*val, label='wfm')
        plt.plot(1e6*tim[sel], 1e3*val[sel], label='sel')
        plt.plot(1e6*tim, 1e3*val_fit, label='bckgrd')
        plt.plot(1e6*tim_fix, 1e3*val_fix, label='fix')
        plt.xlabel('time [µs]')
        plt.ylabel('Voltage [mV]')
        plt.grid()
        plt.legend()
        plt.tight_layout()
        plt.show()

        print(f'charge   : {charge:.3f} nC')
        print(f'timepeak : {1e6*tim_peak:.3f} µs')  
        print(f'offset   : {1e3*pcoeffs[0]:.2f} mV')
        print(f'slope    : {1e-3*pcoeffs[1]:.2f} mV/µs')

    return params
```
